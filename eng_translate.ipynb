```python
from datasets import load_dataset
from deep_translator import GoogleTranslator
from tqdm import tqdm
import pandas as pd

dataset = load_dataset("gentaiscool/codemixqa", split="test")

# keep only Hindi
dataset = dataset.filter(lambda x: x["language"] == "Hindi")

translator = GoogleTranslator(source="auto", target="en")

translated_rows = []

for example in tqdm(dataset, desc="Translating Hindi"):

    text = example["problem"]

    try:
        translated = translator.translate(text)
    except:
        translated = ""

    example["problem_en"] = translated
    translated_rows.append(example)

df = pd.DataFrame(translated_rows)

df.to_csv("english/codemixqa_hindi_translated.csv", index=False)
df.to_json("english/codemixqa_hindi_translated.json", orient="records", lines=True)

print("Done")
```
